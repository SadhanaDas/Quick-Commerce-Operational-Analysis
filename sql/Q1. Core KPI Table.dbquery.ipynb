```sql
SELECT
  COUNT(DISTINCT order_id) AS total_orders,
  ROUND(SUM(payment_value), 2) AS GMV,
  ROUND(SUM(payment_value) / COUNT(DISTINCT order_id), 2) AS AOV,
  ROUND(AVG(review_score), 2) AS avg_review_score,
  (
    SELECT
      ROUND(repeat_customers * 100 / customers, 2) AS repeat_purchase_rate
    FROM
      (
        SELECT
          COUNT(DISTINCT
            CASE
              WHEN order_count > 1 THEN customer_unique_id
            END
          ) AS repeat_customers,
          COUNT(DISTINCT customer_unique_id) AS customers
        FROM
          (
            SELECT
              customer_unique_id,
              COUNT(DISTINCT order_id) AS order_count
            FROM
              workspace.default.final_quick_comm_dataset
            GROUP BY
              customer_unique_id
          ) s
      ) t
  ) AS repeat_purchase_rate,
  ROUND(COUNT(
    CASE
      WHEN order_status = 'canceled' THEN order_id
    END
  )
    * 100.0
    / COUNT(DISTINCT order_id),2) AS cancellation_rate
FROM
  workspace.default.final_quick_comm_dataset;
```
