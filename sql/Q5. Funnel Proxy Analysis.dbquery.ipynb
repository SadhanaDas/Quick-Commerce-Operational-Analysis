```sql
with raw as(
SELECT
    COUNT(DISTINCT order_id) AS total_orders,

    COUNT(
        DISTINCT CASE
            WHEN order_approved_at IS NOT NULL
            THEN order_id
        END
    ) AS orders_approved,

    COUNT(
        DISTINCT CASE
            WHEN order_delivered_carrier_date IS NOT NULL
            THEN order_id
        END
    ) AS orders_shipped,

    COUNT(
        DISTINCT CASE
            WHEN order_delivered_customer_date IS NOT NULL
            THEN order_id
        END
    ) AS orders_delivered

FROM workspace.default.final_quick_comm_dataset
)

select 
*,
ROUND(orders_approved*100.00/total_orders,2) as pct_orders_approved,
ROUND(orders_shipped*100.00/total_orders ,2)as pct_orders_shipped,
ROUND(orders_delivered*100.00/total_orders ,2)as pct_orders_delivered,

ROUND(orders_shipped*100.00/orders_approved ,2)as pct_approved_orders_shipped,
ROUND(orders_delivered*100.00/orders_approved,2) as pct_approved_orders_delivered,
ROUND(orders_delivered*100.00/orders_shipped ,2)as pct_shipped_orders_delivered
from raw;
```
